##### 2D Plot

```python
df = pd.read_csv('World Happiness Report 2022.csv')
```

```python
df.head()
```

```python
df = df.head()
```

```python
plt.plot(df['Country'],df['Happiness score'])
```

```python
plt.plot(df['Country'],df['Dystopia (1.83) + residual'])
plt.plot(df['Country'],df['Explained by: GDP per capita'])
```

```python
plt.title('Country with Happiness Score')
plt.plot(df['Country'],df['Happiness score'])
```

```python
plt.title('Country with Happiness Score')
plt.plot(df['Country'],df['Happiness score'])
plt.xlabel('Country')
plt.ylabel('Happiness Score')
```

```python
plt.title('World Happiness Report')
plt.plot(df['Country'],df['Explained by: Healthy life expectancy'],color='green')
plt.plot(df['Country'],df['Explained by: Freedom to make life choices'],color='blue')
plt.xlabel('Country')
plt.ylabel('Measures')
```

```python
plt.title('World Happiness Report')
plt.plot(df['Country'],df['Dystopia (1.83) + residual'],color='green',linestyle='dashed')
plt.plot(df['Country'],df['Explained by: GDP per capita'],color='blue',linestyle='dashdot')
plt.xlabel('Country')
plt.ylabel('Values')
```

```python
plt.title('World Happiness Report')
plt.plot(df['Country'],df['Dystopia (1.83) + residual'],color='green',linestyle='dashed',linewidth=2)
plt.plot(df['Country'],df['Explained by: GDP per capita'],color='blue',linestyle='dashdot',linewidth=4)
plt.xlabel('Country')
plt.ylabel('Values')
```

```python
plt.title('World Happiness Report')
plt.plot(df['Country'],df['Explained by: Healthy life expectancy'],color='green',marker='D')
plt.plot(df['Country'],df['Explained by: Freedom to make life choices'],color='blue',marker='+')
plt.xlabel('Country')
plt.ylabel('Measures')
```

```python
plt.title('World Happiness Report')
plt.plot(df['Country'],df['Explained by: Healthy life expectancy'],color='green',marker='D',markersize=7)
plt.plot(df['Country'],df['Explained by: Freedom to make life choices'],color='blue',marker='+',markersize=10)
plt.xlabel('Country')
plt.ylabel('Measures')
```

```python
plt.title('World Happiness Report')
plt.plot(df['Country'],df['Explained by: Healthy life expectancy'],color='green',marker='D',markersize=7,label='Healthy life expentancy')
plt.plot(df['Country'],df['Explained by: Freedom to make life choices'],color='blue',marker='+',markersize=10,label='Freedom to make life choices')
plt.xlabel('Country')
plt.ylabel('Measures')
plt.legend()
```

```python
plt.title('World Happiness Report')
plt.plot(df['Country'],df['Explained by: Healthy life expectancy'],color='green',marker='D',markersize=7,label='Healthy life expentancy')
plt.plot(df['Country'],df['Explained by: Freedom to make life choices'],color='blue',marker='+',markersize=10,label='Freedom to make life choices')
plt.xlabel('Country')
plt.ylabel('Measures')
plt.legend(loc='upper left')
```

```python
plt.title('World Happiness Report')
plt.plot(df['Country'],df['Explained by: Healthy life expectancy'],color='green',marker='D',markersize=7,label='Healthy life expentancy')
plt.plot(df['Country'],df['Explained by: Freedom to make life choices'],color='blue',marker='+',markersize=10,label='Freedom to make life choices')
plt.xlabel('Country')
plt.ylabel('Measures')
plt.legend(loc='upper left')
plt.ylim(0.65,0.8)
```
