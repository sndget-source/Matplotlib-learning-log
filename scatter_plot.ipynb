### Import libraries

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

### read csv

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

### set size of graph

```python
plt.figure(figsize=(10,8))
```

### simple scatter plot 

```python
plt.scatter(df['Age'],df['Fare'])
```

#### Same above scatter with x,y labels, title, use alpha parameter for opacity/transparency and use plt.show() because it clears the figure afterward, so your next plot doesn't accidentally overlap with the previous one. 

```python
plt.scatter(df['Age'],df['Fare'],alpha=0.5)
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Age vs Fare')
plt.show()
```

#### To display people who survived(green) and who didn't(red) using scatter plot, and plt.legend() displays a small box on the plot showing what each color/label represents.

```python
plt.figure(figsize=(8,6))
survived=df[df['Survived']==1]
not_survived=df[df['Survived']==0]
plt.scatter(survived['Age'], survived['Fare'], color='green', label = 'Survived', alpha=0.6)
plt.scatter(not_survived['Age'],not_survived['Fare'], color='red', label = 'Did Not Survive', alpha=0.6)
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Age vs Fare, colored by Survival')
plt.legend();
plt.show()
```

#### marker, markersize, etc parameters are applicable

```python
plt.figure(figsize=(8,6))
survived=df[df['Survived']==1]
not_survived=df[df['Survived']==0]
plt.scatter(survived['Age'], survived['Fare'], color='green', label = 'Survived', alpha=0.8,marker='D')
plt.scatter(not_survived['Age'],not_survived['Fare'], color='red', label = 'Did Not Survive', alpha=0.5)
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Age vs Fare, colored by Survival')
plt.legend();
plt.show()
```

#### adding a column called Familysize which adds up Sibings/Spouse + Parents/Children + 1(self)

```python
df['FamilySize'] = df['SibSp']+df['Parch']+1
```

#### to make scatter plot represent 3d data like age, fare and familysize through size of bubbles

```python
plt.scatter(df['Age'], df['Fare'], s=df['FamilySize']*30,alpha=0.5)
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Age vs Fare vs FamilySize')
plt.show()
```

#### scatter plot using plt.plot() - faster than plt.scatter()

```python
plt.plot(df['Age'], df['Fare'])
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Age vs Fare')
plt.show()
```
